```python
from sympy import count_roots

from tools.vizualization import plot_length_distribution
```

```python
with open("../data/zh_sent_dataset.tsv") as f:
    sentences = f.readlines()
```

```python
print(sentences[0])
```

```python
# Compute max and average length (in characters) of sentences
def count_words(sentence: str):
    return len(sentence.split())

sentence_lengths = list(map(count_words, sentences))
max_len = max(sentence_lengths)
avg_len = sum(sentence_lengths) / len(sentences)

print(f"Maximum sentence length: {max_len}")
print(f"Average sentence length: {avg_len:.2f}")
```

```python
plot_length_distribution(sentence_lengths, title="Sentence Length Distribution (Characters)")
```

```python
with open("../data/zh_T9_dataset.tsv") as f:
    lines = f.readlines()

lines = [tuple(line.split()) for line in lines]
digits, pinyins, hanzis = zip(*lines)
```

```python
digit_lengths = [len(digit) for digit in digits]
max_len = max(digit_lengths)
avg_len = sum(digit_lengths) / len(digits)

print(f"Maximum digits length: {max_len}")
print(f"Average digits length: {avg_len:.2f}")
```

```python
plot_length_distribution(digit_lengths, title="Digit Sequence Length Distribution")
```
